```python
# Q1.  Array operations and vectorization

import numpy as np
import pandas as pd

arr = np.array([10, 20, 30])
s = pd.Series([10, 20, 30])

print(arr * 2)
print(s * 2)
```

Explanation

- Both NumPy arrays and Pandas Series support vectorized multiplication, so every element is multiplied by 2.

- Similarity: Both perform element-wise operations.
- Difference: A NumPy array mainly stores values, while a Pandas Series also has an index.

```python
# Q1. NumPy Broadcasting
#Two NumPy arrays have shapes (3, 1) and (1, 3). Predict the shape of their element-wise sum and explain how broadcasting works.
import numpy as np

a = np.array([[10], [20], [30]])

b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
```

**1. Predict the output:**

The output is:

[[11 12 13]
 [21 22 23]
 [31 32 33]]

The shape of the result is (3, 3).

**2. Shapes before addition:**

- Array `a` has shape (3, 1).
- Array `b` has shape (1, 3).

**3. How broadcasting works:**

NumPy broadcasts the dimensions of size 1 so that the two arrays become compatible for element-wise addition. Therefore, the arrays produce a final shape of (3, 3).

**4. If b is replaced with np.array([1, 2]):**

The shapes become (3, 1) and (2,). These shapes are broadcast-compatible, so the result will be:

[[11 12]
 [21 32]
 [31 32]]

```python
# Q3. Extreme Salary Value
# A salary column contains the values 25000, 30000, 35000, and 10000000. Explain how the extreme value affects the mean compared with the median.
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())
```

Median = 32,500.0


**2. Identify the effect of the extreme salary on the mean.**

The salary of 10,000,000 is an extreme value. It greatly increases the mean because the mean uses every value in the calculation. The median is much less affected by the extreme value.


**3. Which statistic may better represent a typical employee's salary?**

The median may better represent a typical employee's salary because it is less affected by the extreme salary of 10,000,000. The median salary is 32,500, which is closer to the salaries of employees A, B, and C.

### Section D – Scenario-Based Problems

```python
import pandas as pd
# 1. Load the dataset
df = pd.read_csv("student_assessment_dirty.csv")

# Inspect the dataset
print(df.head())
df.info()
print(df.isnull().sum())

# 2. Standardize Department names
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .replace({
        "I.T.": "IT",
        "FINANCE": "FINANCE"
    })
)

# 3. Remove exact duplicate records
before_duplicates = len(df)
df = df.drop_duplicates()
duplicates_removed = before_duplicates - len(df)

# 4. Preserve original marks for investigation
df["Original_Marks"] = df["Marks"]

# Convert Marks to string first
marks_text = df["Marks"].astype("string").str.strip()

# Identify Absent separately
df["Status"] = marks_text.mask(
    marks_text.isna() | marks_text.isin(["NA", ""]),
    "Missing"
)
df.loc[marks_text.str.lower() == "absent", "Status"] = "Absent"

# Remove "marks" text from values such as "88 marks"
clean_marks = (
    marks_text
    .str.replace("marks", "", case=False, regex=False)
    .str.strip()
)

# Convert to numeric
df["Marks"] = pd.to_numeric(clean_marks, errors="coerce")
```

```python
# 5. Identify invalid marks
invalid_mask = (df["Marks"] < 0) | (df["Marks"] > 100)

invalid_records = df[invalid_mask].copy()

print("Invalid marks:")
print(invalid_records)

# Convert invalid marks to missing
df.loc[invalid_mask, "Marks"] = pd.NA
df.loc[invalid_mask, "Status"] = "Invalid"

```

```python
# Display cleaned DataFrame
print("\nCleaned DataFrame:")
print(df)
```

```python
print("\nDuplicate records removed:", duplicates_removed)
print("Original missing/absent marks:", 
      df["Original_Marks"].isin(["", "NA", "Absent"]).sum())
print("Invalid marks identified:", len(invalid_records))
print("Missing numeric marks remaining:", df["Marks"].isna().sum())
```

### Explanation

The dataset was cleaned by standardizing department names, removing exact duplicate records, converting the Marks column to numeric values, handling missing and absent marks, and identifying marks outside the valid range of 0–100.

The value "88 marks" was converted to 88. "Absent" was kept as a separate status, while blank and "NA" values were treated as missing after investigation.

The mark 105 was identified as invalid because valid marks must be between 0 and 100.

Missing values should not be deleted without investigation because they may have different meanings, such as a student being absent or marks not yet being entered. Deleting them blindly can result in loss of information and potentially misleading analysis.
